## Q7.
Define the Y-combinator, and give a brief explanation on how it works.

```mikrokosmos
fix != (\f.(\x.f (x x)) (\x.f (x x)))
```

here we give fix a function with some kind of terminating condition. If we do a single step of beta reduction we can see that (where $Y$ is fix and $R$ is func (or our function to recurse))

$$
\begin{aligned}
&\text{fix func} \\
&(\lambda f.(\lambda x.f (x x)) (\lambda x.f (x x))) \text{func} : (f \leftarrow \text{func}) \\
&((\lambda x.\text{func} (x x)) (\lambda x. \text{func} (x x))) (\text{notice this is equivelant to fix func or } Y R) \\
&((\lambda x.\text{func} (x x)) (\lambda x. \text{func} (x x))) : (x \leftarrow (\lambda x. \text{func} (x x))) \\
&\text{func} ((\lambda x.\text{func} (x x)) (\lambda x. \text{func} (x x))) \\
&\text{giving us}\\
&\text{func } (\text{ fix func }) 
\end{aligned}
$$

Or in other words, we can see that $Y R \rightarrow R (Y R)$ applying the function to itself recursively

## Q8.
Define a list in lambda calculus; explain how "indexing" works

```mikrokosmos
# Tuple list definition of a list.
nil = \f.I
pair = \a.\b.\f. ((f a) b)

tupleList = pair 5 (pair 4 (pair 3 (pair 2 (pair 1 nil))))

tupleList false false true
# here we have false -> tail and true -> head
```

```mikrokosmos
# Cons definition of a list.
# The interpretation of (cons h t) is cons of h and the interpretation of t 
nil  = \c.\n.n
# notice this nil definition is the same as above
cons = \h.\t.\c.\n.(c h (t c n))

getIndex = \l.\i. head (i tail l)

consList = cons 1 (cons 2 (cons 3 nil))

getIndex consList 2
```

we can notice that these two definitions are quite similar in how we link a head (the actual value) with a tail (the rest of the list). However the two defintions make indexing into them slightly different. For the first one we can put $i$ many false before the true to get the $i$ th index. For the cons definition we can use our getIndex function which allows us to find the index using a function rather than repeating values.

The benefit of the tuple based definition is that it is more intuitive at first for working with the list

The benefit of using the cons definition is that we can now apply common lambda functions onto the list easily. These functions like map and fold and filter and sum make the list much more useful as a data type
